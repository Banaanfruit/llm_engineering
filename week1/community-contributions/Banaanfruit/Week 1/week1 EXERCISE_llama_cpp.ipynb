```python
from openai import OpenAI
from IPython.display import Markdown, display, update_display
from dotenv import load_dotenv
import sys
sys.path.insert(0,'c:\\Users\\teede\\projects\\llm_engineering\\week1')
import requests
openai = OpenAI()


class create_prompt:
    def __init__(self,prompt= "", method='llama_cp', model= 'gpt-4o-mini' ):
        '''Can be used to create prompts either through openai or llama cp'''
        self.method = method
        self.model = model
        self.prompt = prompt

    def create_prompt(self):
        if self.prompt == "":
            USER_PROMPT = input("Please provide a prompt")
        SYSTEM_PROMPT = "You are a helpful assistant and expert programmer, please help the user with their question."

        self.messages = {"role": "system", "content": SYSTEM_PROMPT}, {"role": "user", "content": USER_PROMPT}



    def send_prompt(self):
        self.create_prompt()

        if self.method == 'openai':
            stream = openai.chat.completions.create(
                model=self.model,
                messages=self.messages,
                stream=True, response_format = {'type': 'text'})
            self.response = ""
        
        elif self.method == 'llama_cp':
            llama_cpp = "http://127.0.0.1:8080/v1/"
            self.model = 'mistralai_Ministral-3-14B-Instruct-2512-GGUF_Ministral-3-14B-Instruct-2512-Q4_K_M.gguf'
            client = OpenAI(base_url=llama_cpp, api_key='sk-no-key-needed')
            stream = client.chat.completions.create(
                model=self.model,
                messages=self.messages,
                stream=True, response_format = {'type': 'text'})
            self.response = ""
        
        else:
            raise Exception('Invalid method')
        
        display_handle = display(Markdown(""), display_id=True)
        for chunk in stream:
            self.response += chunk.choices[0].delta.content or ''
            update_display(Markdown(self.response), display_id=display_handle.display_id)
```
